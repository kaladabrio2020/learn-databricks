## Delta Lake

O Delta Lake é uma camada de armazenamento de código aberto (open-source) que traz confiabilidade, desempenho e qualidade aos data lakes (lagos de dados). Criado originalmente pela Databricks, ele é a base para a construção de uma arquitetura Lakehouse — um modelo que combina a flexibilidade e o baixo custo dos data lakes com a estrutura e o gerenciamento de dados de um data warehouse.

Em termos práticos, o Delta Lake fica "em cima" do seu armazenamento existente (como Amazon S3, Azure Data Lake ou Google Cloud Storage) e gerencia como os dados são gravados e lidos.

O problema que ele resolve
Tradicionalmente, os data lakes eram ótimos para armazenar volumes gigantescos de dados baratos, mas péssimos em mantê-los organizados. Se um processo de gravação falhasse no meio, você ficava com dados corrompidos. Se o esquema de um arquivo mudasse, as consultas quebravam. Esse caos transformava os lagos de dados em "pântanos de dados" (data swamps).

O Delta Lake resolve isso adicionando uma camada de controle.

* Principais Recursos do Delta Lake
Transações ACID: Garante que uma operação de gravação seja concluída totalmente ou desfeita totalmente. Se um processo falhar na metade, seus dados não ficarão corrompidos. Múltiplos usuários podem ler e gravar ao mesmo tempo com segurança.

* Viagem no Tempo (Time Travel): Como o Delta Lake mantém um log de todas as alterações feitas, você pode consultar versões anteriores dos seus dados. Isso é essencial para auditorias, reproduzir modelos de Machine Learning ou reverter exclusões acidentais.

* Imposição e Evolução de Esquema (Schema Enforcement/Evolution): Ele bloqueia a entrada de dados que não correspondam à estrutura esperada (evitando sujeira na base), mas permite que você altere o esquema de forma controlada quando o negócio exigir novas colunas.

* Unificação de Lote e Streaming: Em sistemas antigos, você precisava de uma ferramenta para processar dados em lote (arquivos diários) e outra para streaming (dados em tempo real). O Delta Lake permite que você faça ambos na mesma tabela de forma simultânea.

```python
data = spark\
        .read\
        .format("json")\
        .options(
            inferSchema="true",
            header="true",
        )\
        .load("/Volumes/learn_databricks/schema/volume/Ampla/V_OCORRENCIA_AMPLA.json")
```

```python
data.write.format('delta').mode("overwrite").saveAsTable("learn_databricks.default.ampla")
```

```python
%sql
SELECT * FROM learn_databricks.default.ampla
```

```python
%sql
DELETE FROM learn_databricks.default.ampla
WHERE CLS IS NULL
```

### Particionando

```python
data.write.format('delta').partitionBy("UF").mode("overwrite").saveAsTable("learn_databricks.default.ampla_particionado")
```

```python
%sql
DESCRIBE DETAIL learn_databricks.default.ampla_particionado
```

```python
%sql
SELECT      
   *
FROM  
    learn_databricks.default.ampla_particionado
WHERE   
    UF = "MG"
```

## Otimização

```python
%sql

SELECT  
    Aerodromo_de_Origem,
    concat(Regiao, '-', UF) AS cidade_uf,
    REPLACE(Aerodromo_de_Destino, "OI") AS new_col,
    SUM(
        IFNULL(TRY_CAST(Longitude AS float), 1000)
    ) AS sum_longitude,
    COUNT(DISTINCT matricula) AS distinct_matricula,
    COUNT(*) AS total
FROM  
    learn_databricks.default.ampla
GROUP BY    
    ALL
HAVING
    Aerodromo_de_Origem IS NOT NULL
ORDER BY    
    1,2
```
