```python
data = spark\
        .read\
        .format("csv")\
        .options(**{"sep":",","header":True,"inferSchema":True,"encoding":"UTF-8"})\
        .load("/Volumes/learn_databricks/schema/volume/Bike/orders.csv")
data.show(3)
```

```python
data.write.format("delta").saveAsTable("learn_databricks.schema.orders", mode="overwrite", partitionBy="order_date")
```
