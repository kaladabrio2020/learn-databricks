Entender a estrutura de custos do Databricks pode parecer confuso no início porque ela é composta por duas "fatias" principais. Para clarear o cenário, precisamos falar sobre o que é uma **DBU**, como os modelos de computação (**Classic vs. Serverless**) impactam a conta e onde moram os custos ocultos.

---

## 1. O que é uma DBU (Databricks Unit)?

A **DBU** é a unidade de medida do Databricks para quantificar a capacidade de processamento por segundo. Pense nela como os "créditos" que você gasta para rodar seus códigos. O consumo de DBUs varia de acordo com:

* **O tipo de máquina virtual** (uma máquina com mais CPU e memória consome mais DBUs por hora).
* **A categoria de carga de trabalho** (Workload). O Databricks cobra valores diferentes por DBU para fins diferentes:
* **Jobs Compute (Engenharia de Dados/ETL):** Mais barato (ex: ~$0,15 a $0,40 por DBU), pois são processos em lote e automatizados.
* **All-Purpose Compute (Interativo/Data Science):** Mais caro (ex: ~$0,55 a $0,75 por DBU), usado para exploração em Notebooks em tempo real.
* **SQL Compute (Data Warehousing/BI):** Otimizado para consultas SQL rápidas.



---

## 2. Classic Compute vs. Serverless Compute

A grande virada na arquitetura (e nos custos) do Databricks é a escolha entre o modelo tradicional (Classic) e o modelo sem servidor (Serverless).

### Modelo Classic (Tradicional)

Você cria e gerencia o cluster de máquinas virtuais. O cluster roda dentro da sua própria conta de nuvem (AWS, Azure ou GCP).

* **A cobrança é dupla:** Você paga as **DBUs** para o Databricks + a infraestrutura de **Máquinas Virtuais (VMs)** diretamente para o seu provedor de nuvem (AWS/Azure/GCP).
* **Como funciona o custo:** Você paga pelo tempo em que o cluster estiver ligado (*uptime*), mesmo se ele estiver ocioso. Ele leva de 3 a 10 minutos para iniciar.
* **Ideal para:** Pipelines longos (ETL que roda por horas), processamentos previsíveis e pesados em que você consegue tunar as máquinas perfeitamente.

### Modelo Serverless (Sem Servidor)

O Databricks gerencia toda a infraestrutura computacional dentro da própria conta dele. Você não vê ou configura máquinas virtuais.

* **A cobrança é unificada:** Você paga **apenas uma taxa de DBU mais alta** para o Databricks, que já engloba o processamento e a infraestrutura básica. Não há cobrança separada de VMs de computação na sua nuvem.
* **Como funciona o custo:** O início é quase instantâneo (segundos) e você é cobrado estritamente pelo tempo de execução da query ou do Job. Acabou a query, o custo zera imediatamente.
* **Ideal para:** Consultas em ferramentas de BI (PowerBI/Tableau), analistas trabalhando de forma interativa, pipelines curtos (menores que 10-15 minutos) e demandas altamente sazonais.

---

## 3. Comparativo de Custos: Qual escolher?

| Critério | Classic Compute | Serverless Compute |
| --- | --- | --- |
| **Preço por DBU** | Mais baixo | Mais alto (geralmente o dobro ou mais) |
| **Faturamento Adicional** | VMs cobradas à parte pelo Cloud Provider | Embutido na DBU (salvo rede/transferência) |
| **Tempo de Inicialização** | 3 a 10 minutos (você paga por esse aquecimento) | Instantâneo (30 segundos ou menos) |
| **Desperdício (Desoneração)** | Alto, se esquecer o cluster ligado ou se subutilizar | Zero. Pagamento por uso real (*pay-per-use*) |

> ⚠️ **O erro mais comum:** Olhar apenas a tabela de preços e ver que a DBU Serverless é mais cara. Se você tem um Job que roda em 2 minutos, no modelo *Classic* você vai pagar por 5 minutos de inicialização de máquina + 2 minutos de execução. No *Serverless*, você pagará apenas pelos 2 minutos. Nesse cenário, o Serverless sai mais barato.

---

## 4. Onde moram os custos ocultos?

Ao desenhar sua arquitetura no Databricks, fique atento a três fatores que costumam estourar o orçamento:

1. **Transferência de Dados (Egress/Data Transfer):** Mover dados entre regiões diferentes da nuvem ou tirar dados da nuvem para o ambiente local gera cobrança de rede. No Serverless, o tráfego entre a conta do Databricks e o seu storage de dados pode incorrer em custos pequenos de rede, dependendo de como os endpoints privados são configurados.
2. **Armazenamento (Storage):** O Databricks em si não guarda os dados pesados (eles ficam no seu Data Lake: S3, ADLS ou GCS). No entanto, metadados, tabelas gerenciadas pelo Unity Catalog e logs ocupam espaço e geram uma cobrança residual (DSUs - Databricks Storage Units).
3. **Plano de Nível da Conta (Tiers):** O Databricks possui os planos *Standard* e *Premium/Enterprise*. O plano Premium adiciona governança avançada (Unity Catalog) e segurança, mas aumenta o preço base de **todas** as suas DBUs em aproximadamente 1.5x.

### Resumo prático para economizar:

* Para **BI e analistas** disparando queries esporádicas: Vá de **Serverless**.
* Para **ETL pesado que roda de madrugada** por mais de 30-40 minutos seguidos: O modelo **Classic** (com políticas de auto-terminate agressivas) costuma ser financeiramente mais vantajoso.
