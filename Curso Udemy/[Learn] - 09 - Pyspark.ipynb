```python
from pyspark.sql import functions as F
from pyspark.sql import types as T
```

```python
options_dct = dict(
    inferschema = True,
    header = True
)
path_ds = "/Volumes/learn_databricks/schema/volume/Ampla/V_OCORRENCIA_AMPLA.json"
```

## Filtro de dados (`filter` | `where`)

```python
data.filter(F.col("UF")=="SP").display()
```

Filtro composto

> Não existe nenhuma diferença entre filter() e where() no PySpark. Eles executam exatamente a mesma operação, possuem o mesmo desempenho e são tratados como sinônimos diretos pelo otimizador do Spark. A escolha entre um e outro depende apenas do seu estilo de programação.

```python
data.where(
    (F.col("UF") ==  "SP") | (F.col("Tipo_de_Aerodromo").contains("Púb"))
).limit(3).display()
```

### Functions(funções)

```python
data.select(
    F.lower(F.col("UF"))
).distinct().show()
```

```python
data.select(
    F.when(F.col("UF") == "SP", "Is SP").otherwise("Is Not SP").alias("punto")
).groupBy("punto").count().show()
```

### UDF

As UDFs padrão operam linha por linha. O Spark pega o dado na JVM (Java Virtual Machine), converte (serializa) para o Python processar, e depois converte de volta.

```python
@F.udf(returnType=F.StringType())
def is_sp(x):
    return "Is SP" if x == "SP" else "Is Not SP"
data.select(
    is_sp(F.col("UF")).alias("punto")
).groupBy("punto").count().show()
```

O grande problema das UDFs tradicionais é a lentidão causada pela serialização linha a linha. Para resolver isso, o PySpark introduziu as Pandas UDFs.

Elas usam o Apache Arrow para transferir dados em blocos (batches) em vez de linha por linha. O Pandas executa a operação em toda a série de uma vez, o que pode ser até 100x mais rápido.

```python
import pandas as pd
```

```python
@F.pandas_udf(returnType=F.StringType())
def is_sp_pandas(x):
    return pd.Series(["Is SP" if i == "SP" else "Is Not SP" for i in x])
data.select(
    is_sp_pandas(F.col("UF")).alias("punto")
).groupBy("punto").count().show()
```

* Evite UDFs sempre que possível: Antes de escrever uma UDF, procure exaustivamente no módulo pyspark.sql.functions. As funções nativas do Spark (escritas em Scala/Java) são otimizadas pelo Catalyst Optimizer e rodam muito mais rápido.

* Tratamento de Nulos: O PySpark pode passar valores nulos (None) para sua função. Sempre garanta que sua UDF trate dados ausentes para evitar que o Job quebre no meio do processamento.

* Sempre prefira Pandas UDFs: Se você realmente precisar usar uma UDF, tente escrevê-la como uma Pandas UDF vetorizada.
